# Activity 4.2
### **Refer to this documentation on the string method in Python https://docs.python.org/2.5/lib/string-methods.html**
## Modify the program below to demonstrate the following:
1. Prompt the user to enter a string.
2. Add if statements to check if the characters are all uppercase or lowercase.

```python
# Activity 4.2
# This program demonstrates several string testing methods.

def main():
    # Get a string from the user.
    user_string = input('Enter a string: ')

    print('This is what I found about that string:')

    # Test the string.
    if user_string.isalnum():
        print('The string is alphanumeric.')

    if user_string.isupper():
        print('The string contains only uppercase characters.')

    if user_string.islower():
        print('The string contains only lowercase characters.')

    if user_string.isdigit():
        print('The string contains only digits.')

    if user_string.isalpha():
        print('The string contains only alphabetic characters.')

    if user_string.isspace():
        print('The string contains only whitespace characters.')

# Call the main function.
main()
```

## Modify the program below to demonstrate the following:
1. Prompt user for a string return the lowercase and uppercase version of the string
2. Prompt the user for a second string.
3. Combines the two strings and prints them together in uppercase and lowercase.

```python
# Activity 4.3
# This program demonstrates string manipulation.

def main():
    # Get a string from the user.
    string1 = input('Enter a string: ')

    # Display lowercase and uppercase versions.
    print('Lowercase:', string1.lower())
    print('Uppercase:', string1.upper())

    # Get another string from the user.
    string2 = input('Enter a second string: ')

    # Combine both strings.
    combined_string = string1 + ' ' + string2

    # Display both strings in uppercase and lowercase.
    print('Combined uppercase:', combined_string.upper())
    print('Combined lowercase:', combined_string.lower())

# Call the main function.
main()
```
