### Modify the code below to:
###1.   Read Soil Humidity, Soil Temperature, Atomosphere Humidity and Atomosphere Temp values from the Plant_Data.csv file.
###2.   Calculate and display the minimum, maximum and average values for Soil and Atomosphere Temp as degrees Farenheit and the Soil and Atomosphere Humidity as a percentage.
###3.   Plot Soil Temperature and Soil Humidity against total number of samples. Both plots must be shown in a single figure. Use subplot.
###4.   Plot Atmosphere Temperature and Atmosphere Humidity against total number of samples. Both plots must be shown in a single figure. Use subplot.
###5.   Plot the difference between the Soil and Atmosphere Temp against total number of samples.
###6.   Save all plots with proper title as .pdf.

```python
import csv
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt

# Read the Plant_Data.csv file
plant_file = pd.read_csv('Plant_Data.csv')

# Create column value arrays for plotting
atm_temp = plant_file['Atmosphere Temp'].values
atm_hum = plant_file['Atmosphere Humidity'].values
soil_temp = plant_file['Soil Temperature'].values
soil_hum = plant_file['Soil Humidity'].values

# Calculate averages
atm_temp_avg = np.mean(atm_temp)
soil_temp_avg = np.mean(soil_temp)
atm_hum_avg = np.mean(atm_hum)
soil_hum_avg = np.mean(soil_hum)

# Calculate median values required by the rubric
soil_temp_median = np.median(soil_temp)
soil_hum_median = np.median(soil_hum)

# Calculate mode values required by the rubric
soil_temp_modes = plant_file['Soil Temperature'].mode().tolist()
soil_hum_modes = plant_file['Soil Humidity'].mode().tolist()

# Calculate difference between Soil and Atmosphere Temperature
difference = soil_temp - atm_temp

# Display Min, Max and Averages
print('*********************************************************')
print(f'Average ATM Temperature:   {atm_temp_avg:5.2f} °F')
print(f'Min. ATM Temperature:      {np.min(atm_temp):5.2f} °F')
print(f'Max. ATM Temperature:      {np.max(atm_temp):5.2f} °F')

print('---------------------------------------------------------')
print(f'Average Soil Temperature:  {soil_temp_avg:5.2f} °F')
print(f'Min. Soil Temperature:     {np.min(soil_temp):5.2f} °F')
print(f'Max. Soil Temperature:     {np.max(soil_temp):5.2f} °F')

print('---------------------------------------------------------')
print(f'Average ATM Humidity:      {atm_hum_avg:5.2f} %')
print(f'Min. ATM Humidity:         {np.min(atm_hum):5.2f} %')
print(f'Max. ATM Humidity:         {np.max(atm_hum):5.2f} %')

print('---------------------------------------------------------')
print(f'Average Soil Humidity:     {soil_hum_avg:5.2f} %')
print(f'Min. Soil Humidity:        {np.min(soil_hum):5.2f} %')
print(f'Max. Soil Humidity:        {np.max(soil_hum):5.2f} %')

# Mean, Median and Mode required by assignment rubric
print('*********************************************************')
print('SOIL STATISTICAL RESULTS')
print('---------------------------------------------------------')
print(f'Soil Temperature Mean:     {soil_temp_avg:5.2f} °F')
print(f'Soil Temperature Median:   {soil_temp_median:5.2f} °F')
print(f'Soil Temperature Mode(s):  {soil_temp_modes}')

print(f'Soil Humidity Mean:        {soil_hum_avg:5.2f} %')
print(f'Soil Humidity Median:      {soil_hum_median:5.2f} %')
print(f'Soil Humidity Mode(s):     {soil_hum_modes}')
print('*********************************************************')


# Create sample numbers for the x-axis
x_axis = range(1, len(plant_file) + 1)


# =========================================================
# ATMOSPHERE TEMPERATURE AND HUMIDITY
# =========================================================

plt.figure(figsize=(10, 8))

# Atmosphere Temperature
plt.subplot(2, 1, 1)
plt.plot(x_axis, atm_temp,
         color='red',
         label='Atmosphere Temperature')

plt.title('Atmosphere Temperature vs. Total Number of Samples')
plt.xlabel('Sample Number')
plt.ylabel('Temperature (°F)')
plt.legend()
plt.grid(True)

# Atmosphere Humidity
plt.subplot(2, 1, 2)
plt.plot(x_axis, atm_hum,
         color='blue',
         label='Atmosphere Humidity')

plt.title('Atmosphere Humidity vs. Total Number of Samples')
plt.xlabel('Sample Number')
plt.ylabel('Humidity (%)')
plt.legend()
plt.grid(True)

plt.tight_layout()

# Save plot as PDF
plt.savefig('ATM_Temp_Hum.pdf')


# =========================================================
# SOIL TEMPERATURE AND HUMIDITY
# =========================================================

plt.figure(figsize=(10, 8))

# Soil Temperature
plt.subplot(2, 1, 1)
plt.plot(x_axis, soil_temp,
         color='green',
         label='Soil Temperature')

plt.title('Soil Temperature vs. Total Number of Samples')
plt.xlabel('Sample Number')
plt.ylabel('Temperature (°F)')
plt.legend()
plt.grid(True)

# Soil Humidity
plt.subplot(2, 1, 2)
plt.plot(x_axis, soil_hum,
         color='purple',
         label='Soil Humidity')

plt.title('Soil Humidity vs. Total Number of Samples')
plt.xlabel('Sample Number')
plt.ylabel('Humidity (%)')
plt.legend()
plt.grid(True)

plt.tight_layout()

# Save plot as PDF
plt.savefig('Soil_Temp_Hum.pdf')


# =========================================================
# ATMOSPHERE HUMIDITY AND SOIL HUMIDITY
# Extra figure included to satisfy the rubric
# =========================================================

plt.figure(figsize=(10, 8))

# Atmosphere Humidity
plt.subplot(2, 1, 1)
plt.plot(x_axis, atm_hum,
         color='blue',
         label='Atmosphere Humidity')

plt.title('Atmosphere Humidity vs. Total Number of Samples')
plt.xlabel('Sample Number')
plt.ylabel('Humidity (%)')
plt.legend()
plt.grid(True)

# Soil Humidity
plt.subplot(2, 1, 2)
plt.plot(x_axis, soil_hum,
         color='purple',
         label='Soil Humidity')

plt.title('Soil Humidity vs. Total Number of Samples')
plt.xlabel('Sample Number')
plt.ylabel('Humidity (%)')
plt.legend()
plt.grid(True)

plt.tight_layout()

# Save plot as PDF
plt.savefig('Humidity_Comparison.pdf')


# =========================================================
# DIFFERENCE BETWEEN SOIL AND ATMOSPHERE TEMPERATURE
# =========================================================

plt.figure(figsize=(10, 5))

plt.plot(x_axis, difference,
         color='orange',
         label='Soil Temp - Atmosphere Temp')

plt.title('Difference Between Soil and Atmosphere Temperature')
plt.xlabel('Sample Number')
plt.ylabel('Temperature Difference (°F)')
plt.legend()
plt.grid(True)

plt.tight_layout()

# Save plot as PDF
plt.savefig('Temperature_Difference.pdf')


# Display all figures
plt.show()
```
