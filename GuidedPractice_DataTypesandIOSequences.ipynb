# Activity 1.2
## Modify the following program to demonstrate the following:

1. Display the data in one line as: Name, age, income
2. Display the age in terms of total number of months
3. Convert the income into EU currency. Assume an exchange rate of 1$ = 0.875EU


```python
# Activity 1.2
# Get the user's name, age, and income.
name = input('What is your name? ')
age = int(input('What is your age? '))
income = float(input('What is your income? '))

# Display the data.
print('Here is the data you entered:')
print('Name:', name)
print('Age:', age)
print(f'Income: ${income:.2f}')

# Convert Age to Months
age_months = age * 12


# Convert US dollars to EU currancy
eu = 0.875
eu_currancy = income * eu

# Display name age income in one line
print(f'Name: {name} Age: {age} Income: ${income:.2f}')

# Display age in months
print('Age In Months: ',age_months)

# Display income in EU
print(f'EU Currency: ${eu_currancy:.2f}')
```

# Activity 1.3
## Modify the following program to demonstrate the following:

1. Ask user to enter monthly tips and then display annual income including tips
2. Calculate the amount taxed on pay and tips. \\
Assume a flat rate of 3.25% (0.0325) tax on pay and 5.75% (0.0575) tax on tips
3. Display the total salary in terms of dollars and cents (round to 2 decimal places)

```python
# Activity 1.3
# This program demonstrates two ways
# a floating-point number can be displayed
# as currency.

monthly_pay = 100.34

# Tax percentage variables
pay_tax_percent = 0.0325
tips_tax_percent = 0.0575

# Ask the user for monthly tips
monthly_tips = float(input('Enter your monthly tips: $'))

# Convert monthly pay and tips to annual amounts
annual_pay = monthly_pay * 12
annual_tips = monthly_tips * 12

# Calculate total annual income before taxes
gross_annual_salary = annual_pay + annual_tips

# Calculate taxes on pay and tips separately
pay_tax = annual_pay * pay_tax_percent
tips_tax = annual_tips * tips_tax_percent
total_tax = pay_tax + tips_tax

# Calculate annual salary after taxes
net_annual_salary = gross_annual_salary - total_tax

# Display the results
print(f'\nAnnual pay: ${annual_pay:,.2f}')
print(f'Annual tips: ${annual_tips:,.2f}')
print(f'Total annual income: ${gross_annual_salary:,.2f}')
print(f'Tax on pay: ${pay_tax:,.2f}')
print(f'Tax on tips: ${tips_tax:,.2f}')
print(f'Total taxes: ${total_tax:,.2f}')
print(f'Total salary after taxes: ${net_annual_salary:,.2f}')
```

# Activity 1.4
## Modify the following program to demonstrate the following:
1. Ask user to enter three more test scores
2. Calculate the total average of all scores
3. Display using format keyword and f string rounded to 2 decimal places

```python
# Activity 1.4
# Ask the user to enter six test scores.

test1 = float(input('Enter the first test score: '))
test2 = float(input('Enter the second test score: '))
test3 = float(input('Enter the third test score: '))
test4 = float(input('Enter the fourth test score: '))
test5 = float(input('Enter the fifth test score: '))
test6 = float(input('Enter the sixth test score: '))

# Calculate the total of all six scores.
total = test1 + test2 + test3 + test4 + test5 + test6

# Calculate the average of all six scores.
average = total / 6

# Display using the format() function.
print('The average score using format is: {:.2f}'.format(average))

# Display using an f-string.
print(f'The average score using an f-string is: {average:.2f}')
```

# Activity 1.5
## Modify the program below to demonstrate the following:
1. Modify the code so the user enters the present value, annual interest rate and number of years vested
2. Display the future value of vested money based on user entered present value as a monetary value (i.e. $1,234.43)

```python
# Activity 1.6
# Calculate the future value of an investment.

# Get the present value.
present_value = float(input('Enter the present value: $'))

# Get the annual interest rate as a percentage.
annual_rate = float(input('Enter the annual interest rate: '))

# Get the number of years invested.
years = int(input('Enter the number of years invested: '))

# Convert the percentage to a decimal.
rate = annual_rate / 100

# Calculate the future value.
future_value = present_value * (1 + rate) ** years

# Display the future value as currency.
print(f'The future value of your investment is: ${future_value:,.2f}')
```

# Activity 1.6
## Modify the program below to demonstrate the following:
1. User enters total number of hours and minutes elapsed
2. Display the total time elapsed in number of seconds

```python
# Activity 1.6
# Convert hours and minutes into total seconds.

# Get the number of hours from the user.
hours = int(input('Enter the number of hours elapsed: '))

# Get the number of minutes from the user.
minutes = int(input('Enter the number of minutes elapsed: '))

# Convert hours and minutes to seconds.
hours_in_seconds = hours * 3600
minutes_in_seconds = minutes * 60

# Calculate the total elapsed time in seconds.
total_seconds = hours_in_seconds + minutes_in_seconds

# Display the result.
print(f'The total time elapsed is {total_seconds} seconds.')
```

## Modify the program below to demonstrate the following:
1. User enters their monthly pay and monthly tax amount amounts
2. Display users annual pay as a gross income figure (pay plus taxes)

```python
# Activity 1.8
# Calculate and display annual income.

monthly_pay = 5000.0
annual_pay = monthly_pay * 12
print(f'Your annual pay is ${annual_pay:,.2f}')

# Ask the user for monthly income and tax rate.
user_monthly = float(input('Enter your monthly pay: $'))
user_tax = float(input('Enter your tax rate percentage: '))

# Convert the tax percentage to a decimal.
user_tax /= 100

# Calculate annual pay.
user_annual = user_monthly * 12

# Calculate monthly tax and annual tax.
user_monthly_tax = user_monthly * user_tax
user_annual_tax = user_monthly_tax * 12

# Calculate gross annual income as directed.
user_gross = user_annual + user_annual_tax

# Display gross annual income.
print(f'Your annual pay is ${user_annual:,.2f}')
print(f'Your annual tax amount is ${user_annual_tax:,.2f}')
print(f'Your gross pay is ${user_gross:,.2f}')
```
