## Modify the program to demonstrate the following:
1. Run the program by downloading the sales.txt file and uploading it in the same colab notebook location as this notebook.
2. Add 10 random values to the sales.txt file such as 55.42221 99.20890 44.222 etc.

```python
# Activity 3.1
# This program uses the for loop to read
# all of the values in the sales.txt file.

import random

def main():
    # Open the sales.txt file for reading.
    sales_file = open('/content/sales.txt', 'r')

    # Read all the lines from the file.
    for line in sales_file:
        # Convert line to a float.
        amount = float(line)
        # Format and display the amount.
        print(f'{amount:.2f}')

    # Close the file.
    sales_file.close()

    # Open the sales.txt file for appending.
    sales_file = open('/content/sales.txt', 'a')

    # Add 10 random values to the file.
    for count in range(10):
        amount = random.uniform(1, 100)
        sales_file.write(f'\n{amount}')

    # Close the file.
    sales_file.close()

# Call the main function.
main()
```

```python
# Activity 3.3
# This program writes five names to a file.
def main():
    # Open a new file.
    outfile = open("my_name2.txt", 'w')

    # Prompt the user for 5 names and ages.
    for count in range(5):
        name = input('Enter a name: ')
        age = input('Enter their age: ')

        # Write the name and age to the file.
        outfile.write(f'{name}, Age: {age}\n')

    # Close the file.
    outfile.close()

# Call the main function.
main()
```

```python
# Activity 3.6
# This program reads all of the values in the newly created .csv file and plots them using matplotlib
# It then adds labels for the x and y axis and a title for the plot as well as a legend and makes each of the individual plots a different color

# Import matplotlib and pandas
from matplotlib import pyplot as plt
import pandas as pd

# Read the data from each of the columns
data = pd.read_csv('plot_me.csv')

# Plot the data from column 1
plt.plot(data['A'], label='Column A', color='red')
plt.plot(data['B'], label='Column B', color='blue')
plt.plot(data['C'], label='Column C', color='green')

# Add labels for the x and y axis
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')

# Turn the grid on
plt.grid(True)

# Add a title for the plot
plt.title('Random Data Plot')

# Add a legend and show the plot
plt.legend()
plt.show()
```

# Activity 3.7
## Modify the program to demonstrate the following:
    
1. Read the data from the .csv file and plot each column individually.
2. Make a subplot that contains ***3 different colored scatter*** plots.
3. The first will be made up of column A and B.
4. The second will be made up of column B and C.
5. The final column will be made up of C and A.
6. Add appropriate titles, legends and x and y axis labels.

```python
# Activity 3.7 Solution
# This program reads all of the values in the newly created .csv file and plots them in a subplot as scatter plots using matplotlib

# Import matplotlib and pandas
from matplotlib import pyplot as plt
import pandas as pd

# Read the data from each of the columns
data = pd.read_csv('plot_me.csv')

# Create a subplot with 1 row and 3 columns
plt.figure(figsize=(15, 5))

# Plot the data from column A vs B add labels for the x and y axis and a title for the plot and a legend
plt.subplot(1, 3, 1)
plt.scatter(data['A'], data['B'], label='Column A vs Column B', color='red')
plt.title('Columns A and B')
plt.xlabel('Column A')
plt.ylabel('Column B')
plt.legend()

# Plot the data from column B vs C add labels for the x and y axis and a title for the plot and a legend
plt.subplot(1, 3, 2)
plt.scatter(data['B'], data['C'], label='Column B vs Column C', color='blue')
plt.title('Columns B and C')
plt.xlabel('Column B')
plt.ylabel('Column C')
plt.legend()

# Plot the data from column C vs A add labels for the x and y axis and a title for the plot and a legend
plt.subplot(1, 3, 3)
plt.scatter(data['C'], data['A'], label='Column C vs Column A', color='green')
plt.title('Columns C and A')
plt.xlabel('Column C')
plt.ylabel('Column A')
plt.legend()

# Display the plots
plt.tight_layout()
plt.show()
```
