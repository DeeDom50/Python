Q1: Write a Python program that prompts user for five grocery items (e.g. apples, flour, oranges, etc.) For each item, the program must ask the user its corresponding price. The items and their prices are written into a file. Please include your code and your text file generated by the program in the screenshots.

````python

def main():
    # Open a file for writing.
    outfile = open('grocery_items.txt', 'w')

    # Ask the user for 5 grocery items and their prices.
    for count in range(5):
        item = input('Enter a grocery item: ')
        price = float(input('Enter the price: $'))

        # Write the item and price to the file.
        outfile.write(f'{item}: ${price:.2f}\n')

    # Close the file.
    outfile.close()

main()
````

Q1: Create a text file with the data given below using notepad or any other editor. Save the file as data .txt. Write a program that will open the file and read the age of each individual and their name and display it on the command
prompt. After reading the entire file, the program must also output the total sum of ages included in the file.

John Adams             25

Patrick Henry          30

Thomas Jefferson       19

George Washington      28

Alexander Hamilton     15

````python
def main():
    # Open the file for reading.
    infile = open('data.txt', 'r')

    # Set the total age to 0.
    total_age = 0

    # Read each line in the file.
    for line in infile:
        data = line.split()

        # Get the first name, last name, and age.
        first_name = data[0]
        last_name = data[1]
        age = int(data[2])

        # Display the name and age.
        print(first_name, last_name, age)

        # Add the age to the total.
        total_age += age

    # Close the file.
    infile.close()

    # Display the total of all ages.
    print('Total age:', total_age)

main()
````

Q2: Write a program that does the following:
> 1. checks the current time using python function datetime.now()
2. Convert the timestamp into hours, minutes, and seconds
3. Copy that time stamp into a file
4. Save the time stamp in a tuple
5. Then add a five second delay in the python program using time.sleep(delay in seconds) function
6. Repeat the above steps to get five time stamp values.
7. Display the values of the tuple on command line

You will need to import the following libraries
```
import time
from datetime import datetime
```

````python
import time
from datetime import datetime

def main():
    # Create an empty tuple to store the timestamps.
    timestamps = ()

    # Open a file for writing.
    outfile = open('timestamps.txt', 'w')

    # Get 5 timestamps.
    for count in range(5):
        current_time = datetime.now()

        # Convert the timestamp into hours, minutes, and seconds.
        timestamp = (current_time.hour,
                     current_time.minute,
                     current_time.second)

        # Add the timestamp to the tuple.
        timestamps += (timestamp,)

        # Write the timestamp to the file.
        outfile.write(f'{timestamp[0]:02d}:{timestamp[1]:02d}:{timestamp[2]:02d}\n')

        # Wait 5 seconds before getting the next timestamp.
        if count < 4:
            time.sleep(5)

    # Close the file.
    outfile.close()

    # Display the timestamp values.
    print('Timestamp values:')
    for timestamp in timestamps:
        print(timestamp)

main()
````

Q3: Write a Python program that lets the user enter the total rainfall for each of 12 months into a list. The program should calculate and display the total rainfall for the year, the average monthly rainfall, the months with the highest and lowest amounts.


````python

def main():
    months = ['January', 'February', 'March', 'April', 'May', 'June',
              'July', 'August', 'September', 'October', 'November', 'December']

    rainfall = []

    # Ask the user for rainfall for each month.
    for month in months:
        amount = float(input(f'Enter rainfall for {month}: '))
        rainfall.append(amount)

    # Calculate total and average rainfall.
    total_rainfall = sum(rainfall)
    average_rainfall = total_rainfall / len(rainfall)

    # Find the highest and lowest rainfall amounts.
    highest = max(rainfall)
    lowest = min(rainfall)

    # Find which months had the highest and lowest rainfall.
    highest_month = months[rainfall.index(highest)]
    lowest_month = months[rainfall.index(lowest)]

    # Display the results.
    print(f'\nTotal rainfall for the year: {total_rainfall:.2f}')
    print(f'Average monthly rainfall: {average_rainfall:.2f}')
    print(f'Highest rainfall: {highest_month} with {highest:.2f}')
    print(f'Lowest rainfall: {lowest_month} with {lowest:.2f}')

main()
````

Q4: Create a tuple of with two integer values, two strings and one character. Provide code for the following:
> 1. Slice the tuple: Displays first 3 elements of the tuple
2. Prints the values in the tuple in reverse
3. Displays the length of the tuple
4. Duplicates the tuple elements into a list
5. Remove one item from the tuple (tuples are immutable so you may need to use indirect ways to do this)
6. Create another tuple called b of ten items and combine b with the original tuple to develop a list

````python


def main():
    # Create a tuple with two integers, two strings, and one character.
    my_tuple = (10, 20, 'Hello', 'Python', 'A')

    # 1. Display the first 3 elements of the tuple.
    print('First 3 elements:', my_tuple[:3])

    # 2. Print the tuple values in reverse.
    print('Tuple in reverse:', my_tuple[::-1])

    # 3. Display the length of the tuple.
    print('Length of tuple:', len(my_tuple))

    # 4. Duplicate the tuple elements into a list.
    duplicated_list = list(my_tuple) * 2
    print('Duplicated list:', duplicated_list)

    # 5. Remove one item from the tuple.
    temp_list = list(my_tuple)
    temp_list.remove(20)
    my_tuple = tuple(temp_list)
    print('Tuple after removing one item:', my_tuple)

    # 6. Create another tuple with 6 items and combine it with the original.
    another_tuple = (30, 40, 'Apple', 'Orange', 'B', 50)
    combined_list = list(my_tuple + another_tuple)
    print('Combined list:', combined_list)

main()
````
