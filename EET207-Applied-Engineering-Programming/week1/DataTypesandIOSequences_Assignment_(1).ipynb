Q1: Write a Python program that prompts user for temperature in Celsius and outputs the value in Fahrenheit.

Formula: Fahrenheit = (Celsius x 95 ) + 32

```python
# Ask the user to enter the temperature in Celsius.
celsius = float(input('Enter the temperature in Celsius: '))

# Convert Celsius to Fahrenheit.
fahrenheit = (celsius * 9 / 5) + 32

# Display the result.
print(f'{celsius:.2f} degrees Celsius is {fahrenheit:.2f} degrees Fahrenheit.')
```

Q2: Write a Python program that prompts the user for their birth year and then calculate and output their age

```python
# Ask the user for their birth year.
birth_year = int(input('Enter your birth year: '))

# Store the current year.
current_year = 2026

# Calculate the user's age.
age = current_year - birth_year

# Display the result.
print(f'You are approximately {age} years old.')
```

Q3: Write a Python program that prompts the user to input list-prices for three items and the discounts in percentages for each item. Calculate the total list price and the actual price paid by the user in total after discount. Program must also display the total savings in dollar amount.

```python
# Get the list price and discount percentage for item 1.
price1 = float(input('Enter the list price for item 1: $'))
discount1 = float(input('Enter the discount percentage for item 1: '))

# Get the list price and discount percentage for item 2.
price2 = float(input('Enter the list price for item 2: $'))
discount2 = float(input('Enter the discount percentage for item 2: '))

# Get the list price and discount percentage for item 3.
price3 = float(input('Enter the list price for item 3: $'))
discount3 = float(input('Enter the discount percentage for item 3: '))

# Calculate the savings for each item.
savings1 = price1 * (discount1 / 100)
savings2 = price2 * (discount2 / 100)
savings3 = price3 * (discount3 / 100)

# Calculate the price paid for each item after discount.
actual_price1 = price1 - savings1
actual_price2 = price2 - savings2
actual_price3 = price3 - savings3

# Calculate the totals.
total_list_price = price1 + price2 + price3
total_savings = savings1 + savings2 + savings3
total_actual_price = actual_price1 + actual_price2 + actual_price3

# Display the results.
print('\nPurchase Summary')
print(f'Total list price: ${total_list_price:,.2f}')
print(f'Total savings: ${total_savings:,.2f}')
print(f'Total price paid after discounts: ${total_actual_price:,.2f}')
```
