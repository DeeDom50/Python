Q1: Write a Python program that lets the user enter the total rainfall for each of 12 months into a list. The program should calculate and display the total rainfall for the year, the average monthly rainfall, the months with the highest and lowest amounts.

```python

def main():
    months = ['January', 'February', 'March', 'April',
              'May', 'June', 'July', 'August', 'September',
              'October', 'November', 'December']

    rainfall = []

    # Get rainfall for each month
    for month in months:
        amount = float(input(f'Enter rainfall for {month}: '))
        rainfall.append(amount)

    # Calculate total and average rainfall
    total = sum(rainfall)
    average = total / len(rainfall)

    # Find highest and lowest rainfall
    highest = max(rainfall)
    lowest = min(rainfall)

    # Find the months with highest and lowest rainfall
    highest_month = months[rainfall.index(highest)]
    lowest_month = months[rainfall.index(lowest)]

    # Display results
    print('\nRainfall Results')
    print(f'Total rainfall for the year: {total:.2f}')
    print(f'Average monthly rainfall: {average:.2f}')
    print(f'Highest rainfall: {highest_month} with {highest:.2f}')
    print(f'Lowest rainfall: {lowest_month} with {lowest:.2f}')


main()
```

Q2: Write a function that accepts two arguments: a list and a number n. Assume that the list contains numbers. The function should display all of the numbers in the list that are greater than the number n.

```python

def greater_than(numbers, n):
    for value in numbers:
        if value > n:
            print(value)


def main():
    numbers = [5, 12, 3, 18, 7, 25, 10]
    n = 10

    print('Numbers greater than', n, 'are:')
    greater_than(numbers, n)


main()
```

Q3:  Write a program that prompts the user to enter their first, middle and last names. Displays their first, middle and last initials. For example, if the user enters John William Smith, the program should display J. W. S.

```python
def main():
    first_name = input('Enter your first name: ')
    middle_name = input('Enter your middle name: ')
    last_name = input('Enter your last name: ')

    print(first_name[0].upper() + '.',
          middle_name[0].upper() + '.',
          last_name[0].upper() + '.')


main()
```

Q4: Write a program that lets the user enter a string and displays the character that appears most frequently in the string.

```python
def main():
    text = input('Enter a string: ')

    most_frequent = ''
    highest_count = 0

    for character in text:
        count = text.count(character)

        if count > highest_count:
            highest_count = count
            most_frequent = character

    print('The most frequent character is:', most_frequent)


main()
```

Q5: Create an empty dictionary for ten student names. Prompt the user to input student names(keys). Prompt user to input the key for each as their student id number(value). Sort the dictionary based on the keys(ascending and decending) and display the value pairs. Prompt the user to enter a student id number and display the keys of the students(student name) who match the value pair.

```python
def main():
    # Create an empty dictionary
    students = {}

    # Get 10 student names and ID numbers
    for count in range(10):
        print('\nStudent', count + 1)
        name = input('Enter student name: ')
        student_id = input('Enter student ID number: ')

        students[name] = student_id

    # Display dictionary in ascending order by student name
    print('\nStudents in Ascending Order:')
    for name, student_id in sorted(students.items()):
        print(name, ':', student_id)

    # Display dictionary in descending order by student name
    print('\nStudents in Descending Order:')
    for name, student_id in sorted(students.items(), reverse=True):
        print(name, ':', student_id)

    # Search for student by ID number
    search_id = input('\nEnter a student ID number to search for: ')

    found = False

    for name, student_id in students.items():
        if student_id == search_id:
            print('Student:', name)
            found = True

    if found == False:
        print('No student found with that ID number.')


main()
```
