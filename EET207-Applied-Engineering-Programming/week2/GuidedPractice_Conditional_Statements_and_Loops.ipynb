# Activity 2.1
## Modify the program below to demonstrate the following:
1. Calculate the total tax amount using a tax percentage of 40% if the gross pay is greater than 7500 and 20% if it is less than or equal to 7500.
2. Use the calculated tax amount in dollars to subtract from gross pay and display the net pay

```python
# Activity 2.1
#Variables to represent the base hours and
# the overtime multiplier.
base_hours = 40       # Base hours per week
ot_multiplier = 1.5   # Overtime multiplier

# Get the hours worked and the hourly pay rate.
hours = float(input('Enter the number of hours worked: '))
pay_rate = float(input('Enter the hourly pay rate: '))

# Calculate and display the gross pay.
if hours > base_hours:
    # Calculate the gross pay with overtime.
    # First, get the number of overtime hours worked.
    overtime_hours = hours - base_hours

    # Calculate the amount of overtime pay.
    overtime_pay = overtime_hours * pay_rate * ot_multiplier

    # Calculate the gross pay.
    gross_pay = base_hours * pay_rate + overtime_pay
else:
    # Calculate the gross pay without overtime.
    gross_pay = hours * pay_rate

# Display the gross pay.
print('The gross pay is $', format(gross_pay, ',.2f'), sep='')

#Calculate tax
if gross_pay > 7500:
    tax_amount = gross_pay * 0.40
else:
    tax_amount = gross_pay * 0.20

# Calculate net pay.
net_pay = gross_pay - tax_amount

# Display tax amount and net pay.
print('The tax amount is $', format(tax_amount, ',.2f'), sep='')
print('The net pay is $', format(net_pay, ',.2f'), sep='')


```

##Modify the program below to demonstrate the following:
Ask the user to enter:
1. start_speed
2. end_speed
3. increment

```python
# Activity 2.4
# This program converts the speeds 60 kph
# through 130 kph (in 10 kph increments)
# to mph.

start_speed = int(input('Enter the starting speed: '))
end_speed = int(input('Enter the ending speed: '))
increment = int(input('Enter the speed increment: '))
conversion_factor = 0.6214  # Conversion factor

# Print the table headings.
print('KPH\tMPH')
print('--------------')

# Print the speeds.
for kph in range(start_speed, end_speed + 1, increment):
    mph = kph * conversion_factor
    print(kph, '\t', format(mph, '.1f'))
```

```python
# Activity 2.5
# This program assists a technician in the process
# of checking a substance's temperature.

# Get the substance's temperature.
temperature = float(input("Enter the substance's Celsius temperature: "))
max_temp = float(input("Enter the maximum temperature: "))

 # As long as necessary, instruct the user to
# adjust the thermostat.
while temperature <= max_temp:
    print('The temperature is less than or equal to the maximum temperature.')
    print('5 minutes. Then take the temperature')
    print('again and enter it.')
    temperature = float(input('Enter the new Celsius temperature: '))

# Remind the user to check the temperature again
# in 15 minutes.
print('The temperature is above the maximum temperature.')
print('Check it again in 15 minutes.')
```

# Activity 2.6
The program below passes the variable 'value' from the 'main' function to the the change_me function.

Answer this question:
> Is it possible to use the variable 'arg' in the 'main' function? Why or Why not?

No, arg cannot be used in the main() function because arg is a local variable of the change_me() function. It only exists within that function.

```python
# Activity 2.6
# This program demonstrates what happens when you
# change the value of a parameter.

def main():
    value = 9100
    print('The value is', value)
    change_me(value)
    print('Back in main the value is', value)

def change_me(arg):
    print('I am changing the value of arg which I received from main function. Right now arg is', arg)
    arg = 0
    print('Now the value of arg is', arg)

# Call the main  function.
main()
```

# Activity 2.8
1. Prompt the user for the number of ounces
2. Display the equivalent number of cups.

```python
# Activity 2.8
# This program converts cups to fluid ounces.

def main():
    # display the intro screen.
    intro()
    # Get the number of cups.
    ounces_needed = int(input('Enter the number of ounces: '))
    # Convert the cups to ounces.
    cups_in_main = ounces_to_cups(ounces_needed)
    print('Cups value sent by the function is', cups_in_main)

# The intro function displays an introductory screen.
def intro():
    print('This program converts measurements')
    print('in cups to fluid ounces. For your')
    print('reference the formula is:')
    print('    1 cup = 8 fluid ounces')
    print()

# The cups_to_ounces function accepts a number of
# cups and displays the equivalent number of ounces.
def ounces_to_cups(ounces):
    cups = ounces / 8
    return cups

# Call the main function.
main()
```
